````python
from pathlib import Path

ROOT = Path.cwd().parent if Path.cwd().name == "notebooks" else Path.cwd()
DATA = ROOT / "data"

expected = [
    "ovarian_subset.h5ad",
    "ovarian_annotated.h5ad",
    "transcripts_crop.parquet",
    "cell_boundaries_crop.parquet",
    "nucleus_boundaries_crop.parquet",
    "crop_metadata.json",
]
for name in expected:
    p = DATA / name
    flag = "OK  " if p.exists() else "MISSING"
    size = f"{p.stat().st_size/1e6:7.1f} MB" if p.exists() else ""
    print(f"{flag} {name:<34} {size}")
````

## 3. Can you load it and draw something?

If a scatter of cells appears below, you are ready.

````python
import matplotlib.pyplot as plt

adata = sc.read_h5ad(DATA / "ovarian_subset.h5ad")
print(adata)

fig, ax = plt.subplots(figsize=(5, 5))
ax.scatter(*adata.obsm["spatial"].T, s=0.4, c="k", linewidths=0)
ax.set_aspect("equal")
ax.invert_yaxis()          # image convention: y increases downwards
ax.set_xlabel("x (um)"); ax.set_ylabel("y (um)")
ax.set_title(f"{adata.n_obs:,} cells")
plt.show()
````

That black cloud is a piece of ovarian tumour, one dot per segmented cell.
